```python
import math
class HinhCau:
    def __init__(self, ban_kinh):
        if ban_kinh <= 0:
            raise ValueError("ban kinh khong duong")
        self.ban_kinh = ban_kinh
    def dien_tich(self):
        return 4 * math.pi * self.ban_kinh**2
    def the_tich(self):
        return 4/3 * math.pi * self.ban_kinh**3
    def __str__(self):
        return f"hinh cau (r={self.ban_kinh})"
hc1 = HinhCau(2)
print(hc1)
print("dien tich :", hc1.dien_tich())
print("the tich:", hc1.the_tich())
try:
    hc2 = HinhCau(-1)
except ValueError as e:
    print("loi:", e)
```

# bài 4.3

```python
class SoDuKhongDu(Exception):
    pass

class TaiKhoan:
    def __init__(self, chu_tai_khoan, so_du=0):
        if so_du < 0:
            raise ValueError("So du ban dau khong duoc am")
        self.chu_tai_khoan = chu_tai_khoan
        self.so_du = so_du
        self.lich_su = []
    def nap(self, so_tien):
        if so_tien <= 0:
            raise ValueError("So tien nap phai lon hon 0")
        self.so_du += so_tien
        self.lich_su.append(f"Nap {so_tien}")
        assert self.so_du >= 0     
    def rut(self, so_tien):
        if so_tien <= 0:
            raise ValueError("So tien rut phai lon hon 0")
        if so_tien > self.so_du:
            raise SoDuKhongDu("Rut qua so du")
        self.so_du -= so_tien
        self.lich_su.append(f"Rut {so_tien}")
        assert self.so_du >= 0
try:
    tai_khoan_sv = TaiKhoan("Sinh vien", 1000000)
    tai_khoan_sv.nap(500000)
    tai_khoan_sv.rut(200000)
    tai_khoan_sv.rut(2000000)
except ValueError as loi:
    print("Loi:", loi)
except SoDuKhongDu as loi:
    print("Loi:", loi)
print("So du cuoi:", tai_khoan_sv.so_du)
print("Lich su:", tai_khoan_sv.lich_su)
```

# bài 4.5

```python
import math

class HinhHoc:
    def dien_tich(self):
        raise NotImplementedError("Phai cai dat trong lop con")        
    def __str__(self):
        return "Hinh hoc"
class HinhTron(HinhHoc):
    def __init__(self, r):
        self.r = r        
    def dien_tich(self):
        return math.pi * (self.r ** 2)        
    def __str__(self):
        return f"Hinh tron (r={self.r})"
class HinhChuNhat(HinhHoc):
    def __init__(self, a, b):
        self.a = a
        self.b = b       
    def dien_tich(self):
        return self.a * self.b        
    def __str__(self):
        return f"Hinh chu nhat ({self.a}x{self.b})"
class HinhTamGiac(HinhHoc):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c     
    def dien_tich(self):
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(p * (p - self.a) * (p - self.b) * (p - self.c))        
    def __str__(self):
       return f"Hinh tam giac ({self.a}, {self.b}, {self.c})"
hinh_tron = HinhTron(1)
print("Da tao:", hinh_tron, "- Dien tich:", hinh_tron.dien_tich())
hinh_chu_nhat = HinhChuNhat(3, 4)
print("Da tao:", hinh_chu_nhat, "- Dien tich:", hinh_chu_nhat.dien_tich())
hinh_tam_giac = HinhTamGiac(3, 4, 5)
print("Da tao:", hinh_tam_giac, "- Dien tich:", hinh_tam_giac.dien_tich())
danh_sach = [hinh_tron, hinh_chu_nhat, hinh_tam_giac]
danh_sach.sort(key=lambda x: x.dien_tich())
print("\nDanh sach sau khi sap xep tang dan:")
for hinh in danh_sach:
    print(hinh, "- Dien tich:", hinh.dien_tich())
tong_dien_tich = 0
for hinh in danh_sach:
    tong_dien_tich += hinh.dien_tich()   
print("\nTong dien tich la:", tong_dien_tich)
```

```python
import pandas as pd
du_lieu = pd.read_csv('goldprice.csv')
du_lieu['nam'] = du_lieu['Date'].str.slice(0, 4).astype(int)
print("Du lieu 5 dong dau:")
print(du_lieu.head())
thong_ke_nam = du_lieu.groupby('nam').agg(so_thang=('Price', 'count'), gia_trung_binh=('Price', 'mean')).reset_index()
print("\nThong ke trung binh va so thang:")
print(thong_ke_nam.head())
thong_ke_nam = thong_ke_nam.sort_values('nam')
thong_ke_nam['muc_tang_phan_tram'] = thong_ke_nam['gia_trung_binh'].pct_change() * 100
print("\nSau khi tinh phan tram tang so voi nam truoc:")
print(thong_ke_nam[['nam', 'gia_trung_binh', 'muc_tang_phan_tram']].head())
nam_du_thang = thong_ke_nam[thong_ke_nam['so_thang'] == 12]
print("\nNhung nam co du 12 thang:")
print(nam_du_thang[['nam', 'so_thang']].head())
ba_nam_tang_nhat = nam_du_thang.nlargest(3, 'muc_tang_phan_tram')
print("\n--- KET QUA: 3 NAM TANG MANH NHAT ---")
print(ba_nam_tang_nhat[['nam', 'muc_tang_phan_tram']])
nam_giam_nhat = nam_du_thang.nsmallest(1, 'muc_tang_phan_tram')
print("\n--- KET QUA: NAM GIAM MANH NHAT ---")
print(nam_giam_nhat[['nam', 'muc_tang_phan_tram']])
```
